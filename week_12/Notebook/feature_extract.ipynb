```python
from keras.applications.resnet50 import ResNet50
from tensorflow.keras.applications import Xception
from keras.applications.vgg16 import VGG16
import numpy as np
import pandas as pd
import pickle
import cv2
import os
from glob import glob
from pickle import dump, load
from keras.models import Model
import json
```

```python
feature_model = VGG16(include_top=True)

extractor = Model(inputs = feature_model.input,outputs = feature_model.layers[-2].output)
# print(extractor.summary())
```

```python
print(extractor.summary())
```

```python

images_path = '../Data/BNG_9k/images/'
images = glob(images_path+'*.png')
print(len(images))
```

```python
images_features = {}
count = 0
for i in images:
    img = cv2.imread(i)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224,224))

    
    
    img = img.reshape(1,224,224,3)
    pred = extractor.predict(img).reshape(4096,)
        
    img_name = i.split('/')[-1]
    
    images_features[img_name] = pred
    
    count += 1
        
    if count % 100 == 0:
        print(count)
```

```python
with open('../Logs/VGG16_BNG9k/image_features.pickle', "wb" ) as pickle_f:
    pickle.dump(images_features, pickle_f)
```

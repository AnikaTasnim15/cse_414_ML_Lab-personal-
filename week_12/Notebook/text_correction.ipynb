```python
with open('../Data/BNG_9k/Captions.txt', encoding="utf-8") as f:
    content = f.readlines()
# you may also want to remove whitespace characters like `\n` at the end of each line
content = [x.strip() for x in content] 
```

```python
updated_content = []
for text in content:
    image_name = text.split('#')[0]
    text_content = text.split('#')[-1]
    image_number = int(image_name.split('.')[0])
    updated_content.append(str(image_number)+'.png#'+text_content)
```

```python
textfile = open("output.txt", "w", encoding="utf-8")
for element in updated_content:
    textfile.write(element + "\n")
textfile.close()
```
